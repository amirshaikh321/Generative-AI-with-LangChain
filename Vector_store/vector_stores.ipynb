```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document


doc1 = Document(
        page_content=(
            "Virat Kohli is one of the most successful batsmen in IPL history. "
            "He has spent his entire IPL career with Royal Challengers Bangalore. "
            "Known for his consistency, aggressive batting style, and leadership, "
            "Kohli won the Orange Cap in 2016 after scoring a record number of runs."
        ),
        metadata={
            "player_name": "Virat Kohli",
            "team": "Royal Challengers Bangalore",
            "role": "Batsman"
        }
    )

doc2 = Document(
        page_content=(
            "MS Dhoni is a legendary Indian cricketer and the former captain of "
            "Chennai Super Kings. He is famous for his calm leadership, finishing "
            "ability, and sharp wicketkeeping skills. Under his captaincy, CSK "
            "won multiple IPL titles."
        ),
        metadata={
            "player_name": "MS Dhoni",
            "team": "Chennai Super Kings",
            "role": "Wicketkeeper-Batsman"
        }
    )

doc3 = Document(
        page_content=(
            "Rohit Sharma is one of the most successful captains in IPL history. "
            "He has led Mumbai Indians to several IPL championships. Rohit is known "
            "for his elegant batting, ability to score big hundreds, and tactical leadership."
        ),
        metadata={
            "player_name": "Rohit Sharma",
            "team": "Mumbai Indians",
            "role": "Batsman"
        }
    )

doc4 = Document(
        page_content=(
            "Jasprit Bumrah is a premier fast bowler in the Indian Premier League. "
            "Playing for Mumbai Indians, he is known for his deadly yorkers, "
            "economical bowling at the death, and match-winning performances."
        ),
        metadata={
            "player_name": "Jasprit Bumrah",
            "team": "Mumbai Indians",
            "role": "Bowler"
        }
    )

doc5 = Document(
        page_content=(
            "Andre Russell is a powerful all-rounder who represents Kolkata Knight Riders. "
            "He is known for his explosive batting, fast bowling, and athletic fielding. "
            "Russell has changed the outcome of many IPL matches with his aggressive style."
        ),
        metadata={
            "player_name": "Andre Russell",
            "team": "Kolkata Knight Riders",
            "role": "All-Rounder"
        }
    )

print(doc1)
docs = [doc1,doc2,doc3,doc4,doc5]
```

```python
from dotenv import load_dotenv
load_dotenv()

vector_store = Chroma.from_documents(
    embedding = GoogleGenerativeAIEmbeddings(model= 'gemini-embedding-001'),
    persist_directory="chroma_db",
    collection_name="sample",
    documents=docs
)
```
