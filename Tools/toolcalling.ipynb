# Tool creation

```python
from langchain.tools import tool

@tool 
def multiply(a:int, b:int ):
    """this is a tool of multiplication of two numbers"""
    return a*b
```

```python
tool = multiply

result = tool.invoke({"a":12,"b":33})
print(result)
```

# tool binding

```python
tool.args
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash")

llm_with_tool = llm.bind_tools([tool])
```

```python
llm_with_tool.invoke('can you multiply 3 with 6').tool_calls
```

# tool execution

```python
from langchain_core.messages import HumanMessage

query = HumanMessage('can you multiply the 3 with the 45')
messages = [query]
```

```python
result = llm_with_tool.invoke(messages)
messages.append(result)
```

```python
result.tool_calls[0]
```

```python
result = multiply.invoke(
{'name': 'multiply',
 'args': {'a': 3, 'b': 45},
 'id': '3393a3e9-b6dc-422b-8780-318c62522384',
 'type': 'tool_call'})
result
```

```python
messages.append(result)

final_output = llm_with_tool.invoke(messages)
final_output.content
```
